```python
'''
The core demographics table is completed, the next step is to melt the data for the multi select tables.
We accept duplicate id values to allow the full answers to be represented, separating each answer into a single sheet format.
'''

# helper function to automate melting, this function simply applies df.melt
# and then removes nulls, drops the helper column, fixes the index, and reorders by the id column

def melt_group(df, id_col, cols, category_name):
    
    # melting only columns that we need
    melted = df[[id_col] + cols].melt(
        id_vars=id_col,
        var_name='_helper',
        value_name=category_name
    )
    
    # post melt fixes
    melted = melted.dropna(subset=[category_name])
    melted = melted.drop(columns='_helper')
    melted = melted.sort_values(id_col)
    
    return melted.reset_index(drop=True)

# now we get to some of the AI GENERATED BITS 
# I didnt want to write out long lists of relevent columns so I passed my function into AI and had it do it

# melting languages 

lang_cols = ['lang_python', 'lang_r', 'lang_sql', 'lang_c', 'lang_csharp',
             'lang_cpp', 'lang_java', 'lang_javascript', 'lang_bash', 'lang_php',
             'lang_matlab', 'lang_julia', 'lang_go', 'lang_other']
df_languages = melt_group(df, 'id', lang_cols, 'language')

# Visualisation libraries
viz_cols = ['viz_matplotlib', 'viz_seaborn', 'viz_plotly', 'viz_ggplot', 'viz_shiny',
            'viz_d3', 'viz_altair', 'viz_bokeh', 'viz_geoplotlib', 'viz_folium',
            'viz_pygal', 'viz_dygraphs', 'viz_highcharter', 'viz_other']
df_viz = melt_group(df, 'id', viz_cols, 'viz_library')

# ML frameworks
framework_cols = ['framework_sklearn', 'framework_tensorflow', 'framework_keras',
                  'framework_pytorch', 'framework_fastai', 'framework_xgboost',
                  'framework_lightgbm', 'framework_catboost', 'framework_caret',
                  'framework_tidymodels', 'framework_jax', 'framework_pytorch_lightning',
                  'framework_huggingface', 'framework_other']
df_frameworks = melt_group(df, 'id', framework_cols, 'framework')

# ML algorithms
algo_cols = ['algo_linear_logistic_regression', 'algo_decision_trees_random_forests',
             'algo_gradient_boosting', 'algo_bayesian', 'algo_evolutionary',
             'algo_dense_neural_networks', 'algo_cnn', 'algo_gan', 'algo_rnn',
             'algo_transformers', 'algo_autoencoders', 'algo_gnn', 'algo_other']
df_algorithms = melt_group(df, 'id', algo_cols, 'algorithm')

# Course platforms
course_cols = ['course_platform_coursera', 'course_platform_edx', 'course_platform_kaggle',
               'course_platform_datacamp', 'course_platform_fastai', 'course_platform_udacity',
               'course_platform_udemy','course_platform_linkedin', 'course_platform_cloud_cert',
               'course_platform_university', 'course_platform_other']
df_courses = melt_group(df, 'id', course_cols, 'course_platform')

# First helpful platforms
first_helpful_cols = ['first_helpful_university', 'first_helpful_online_courses',
                      'first_helpful_social_media', 'first_helpful_video',
                      'first_helpful_kaggle', 'first_helpful_none', 'first_helpful_other']
df_first_helpful = melt_group(df, 'id', first_helpful_cols, 'first_helpful_platform')

# Work activities
activity_cols = ['activity_data_analysis', 'activity_data_infrastructure', 'activity_ml_prototyping',
                 'activity_ml_service', 'activity_ml_experimentation', 'activity_ml_research',
                 'activity_none', 'activity_other']
df_activities = melt_group(df, 'id', activity_cols, 'work_activity')

# Summary
tables = {
    'demographics': df_demographics,
    'languages': df_languages,
    'viz_libraries': df_viz,
    'frameworks': df_frameworks,
    'algorithms': df_algorithms,
    'course_platforms': df_courses,
    'first_helpful': df_first_helpful,
    'work_activities': df_activities
}

for name, table in tables.items():
    print(f"{name}: {len(table)} rows, {table['id'].nunique()} unique respondents")
```
